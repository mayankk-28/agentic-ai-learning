## 1) A tiny deterministic environment

The agent moves on a short line of 5 positions. Start at 0, goal at 4.
- Action `0`: move left
- Action `1`: move right
- Reward: `+1` at goal, otherwise `-0.01`
- Episode ends at goal or max steps

```python
class TinyLineEnv:
    def __init__(self, size=5, max_steps=10):
        self.size = size
        self.max_steps = max_steps
        self.reset()

    def reset(self):
        self.position = 0
        self.steps = 0
        return self.position

    def step(self, action):
        self.steps += 1
        if action == 1:
            self.position = min(self.position + 1, self.size - 1)
        else:
            self.position = max(self.position - 1, 0)

        done = self.position == self.size - 1 or self.steps >= self.max_steps
        reward = 1.0 if self.position == self.size - 1 else -0.01
        info = {"steps": self.steps}
        return self.position, reward, done, info
```

## 2) A simple rule-based agent

This agent always moves right until it reaches the goal.

```python
class RuleBasedAgent:
    def act(self, obs):
        return 1
```

## 3) Run interaction loop

```python
env = TinyLineEnv(size=5, max_steps=10)
agent = RuleBasedAgent()

num_episodes = 3
for episode in range(1, num_episodes + 1):
    obs = env.reset()
    total_reward = 0.0

    while True:
        action = agent.act(obs)
        obs, reward, done, info = env.step(action)
        total_reward += reward
        if done:
            print(f"Episode {episode}: total_reward={total_reward:.2f}, final_position={obs}, steps={info['steps']}")
            break
```
